### O protocolo de argumento

**Afirmação → previsão → evidência → justificativa → objeção → veredito → revisão.**

- **Afirmação:** frase verificável, delimitada por variável, população e período.
- **Hipótese/previsão:** o que esperamos observar, escrito antes de ver a próxima saída.
- **Evidência:** observação ou cálculo que pode ser conferido; opinião e aparência não bastam.
- **Justificativa:** por que aquela evidência sustenta ou enfraquece a afirmação.
- **Objeção:** explicação concorrente ou limitação que poderia mudar o argumento.
- **Veredito:** **sustentada neste recorte**, **refutada neste recorte** ou **inconclusiva**.
  “Não há evidência suficiente de X” não significa “há evidência de que X é falso”.
- **Revisão:** mudança explícita de conclusão diante de evidência melhor. Rever uma hipótese é mérito.

## Dicionário da aula 01 — visualização e percepção

**Visualização de dados** é a representação de dados por elementos visuais para apoiar uma tarefa.
Uma representação é útil quando ajuda uma pessoa a responder uma pergunta com precisão e esforço
adequados. Beleza, rapidez, precisão e lembrança são critérios diferentes.

**Pipeline:** **dado → transformação → codificação → percepção → cognição → decisão**.
Dado é o registro; transformação é filtrar, calcular ou agregar; codificação associa variáveis a
propriedades visuais; percepção detecta padrões; cognição interpreta esses padrões com conhecimento
prévio; decisão é a ação. No resumo em quatro etapas, transformação acompanha o dado e cognição
acompanha a percepção. Uma falha de coleta não é corrigida apenas mudando a paleta.

**Explorar** é procurar padrões e perguntas; **explicar** é comunicar uma conclusão delimitada a uma
audiência; **registrar** é preservar evidências consultáveis. Uma mesma figura pode atender a mais
de uma função. **Tarefa analítica** é a ação do leitor: localizar um valor, comparar, ordenar,
identificar tendência ou detectar exceção. Tabelas favorecem consultas exatas; gráficos podem
facilitar padrões. A escolha depende da tarefa e do público.

**Percepção pré-atentiva** é a detecção rápida de certos contrastes antes de uma inspeção consciente
detalhada. Exemplos: matiz, intensidade, tamanho, orientação, posição, forma simples, curvatura,
movimento e presença de uma marca adicional/junção. Seu efeito depende do contexto: nem toda
diferença de forma produz um destaque imediato.

**Pop-out** é um destaque que parece “saltar aos olhos”. Em uma **busca por atributo único**, o alvo
difere dos demais por uma característica, como a única marca laranja. Uma **busca por conjunção**
exige combinar atributos, como localizar o quadrado laranja entre círculos laranja e quadrados
azuis. Esta costuma exigir mais atenção. Tempo pouco sensível ao número de distratores é uma
tendência experimental em certas tarefas, não uma garantia para qualquer gráfico.

**Visão foveal** favorece detalhes na região fixada pelo olhar; **visão periférica** favorece a
percepção do entorno e de mudanças, com menor resolução para ler detalhes. **Atenção seletiva**
prioriza parte da cena. **Memória visual de trabalho** mantém uma quantidade limitada de informação
por pouco tempo; a referência didática de cerca de quatro objetos varia com a tarefa e o agrupamento.
Uma legenda distante exige alternar o olhar e memorizar correspondências; rótulos diretos reduzem
esse custo. **Carga cognitiva** é a demanda de processamento imposta pela tarefa e pela apresentação.

**Casos históricos como perguntas de projeto:** o mapa de John Snow associa ocorrências de cólera
ao espaço urbano; a carta de Minard integra localização, direção e perdas da campanha de Napoleão;
o caso Challenger é discutido na disciplina para examinar seleção de evidências e comunicação de risco.
Em cada caso, perguntem o que foi medido, o que foi omitido e qual decisão estava em jogo.
Uma visualização isolada não substitui a investigação histórica ou causal.

## Dicionário da aula 02 — marcas, canais e precisão

**Marca** é a entidade gráfica: ponto (0D), linha (1D), área/superfície (2D) ou volume (3D).
Um ponto desenhado tem tamanho na tela, mas funciona como localização. **Canal** é uma propriedade
da marca que varia para representar dados. Uma barra é uma marca de área, mas pode codificar um
número principalmente pelo comprimento. Não confundam geometria da marca com canal utilizado.

| Canal | O que varia | Uso e cuidado |
|---|---|---|
| Posição horizontal/vertical | Localização em uma escala | Comparação numérica precisa quando há referência comum |
| Comprimento | Extensão de uma marca | Magnitude; exige base apropriada, zero em barras comuns |
| Ângulo e inclinação | Abertura ou direção | Proporção ou mudança; dependem do enquadramento |
| Área e volume | Superfície ou espaço ocupado | Magnitude; diferenças são difíceis de estimar com precisão |
| Curvatura | Quanto uma linha se curva | Pode distinguir formas; comparação quantitativa é difícil |
| Matiz | Família de cor | Identidade, como equipe; não tem ordem numérica natural |
| Saturação/croma | Intensidade cromática | Pode sugerir ordem; baixa precisão para valores exatos |
| Luminosidade | Aparência de claro/escuro | Ordem em escalas sequenciais |
| Forma, textura/grão e orientação | Símbolo, padrão ou direção | Identidade; textura também pode sugerir ordem |
| Movimento | Deslocamento ou mudança no tempo | Atrai atenção e pode sugerir agrupamento |

**Canais de identidade** distinguem “qual”; **canais de magnitude** mostram “quanto” ou “em que
ordem”. **Redundância deliberada** representa a mesma variável por mais de um canal, como cor e
forma para equipe. Ajuda na acessibilidade. Usar canais extras sem tarefa clara pode gerar ruído.

| Tipo de dado | Definição | Exemplo | Codificação compatível |
|---|---|---|---|
| Nominal (N) | Categorias sem ordem intrínseca | Equipe A/B | Posição categórica, forma, matiz |
| Ordinal (O) | Ordem sem distância quantitativa conhecida | Baixa/média/alta prioridade | Posição ordenada, luminosidade |
| Quantitativo (Q) | Valor numérico com diferenças interpretáveis | Quantidade, duração | Posição, comprimento; área se a tarefa permitir |
| Temporal (T) | Instante, intervalo ou ciclo de tempo | Mês, horário | Posição cronológica; cor cíclica para ciclos |

**Escala intervalar** permite comparar diferenças, mas o zero é convencional (temperatura em °C).
**Escala de razão** tem zero significativo e permite razões (duração). 20 °C não é “duas vezes
mais quente” que 10 °C. Um código numérico de categoria não torna a variável quantitativa.

**Hierarquia de Cleveland–McGill:** para certas tarefas de comparação de magnitude, posição em
escala comum costuma permitir maior precisão; depois vêm posição em escalas não alinhadas,
comprimento/direção/ângulo, área, volume/curvatura e sombreamento/saturação. É uma síntese de
resultados experimentais, não uma ordem universal para qualquer tarefa. As replicações de
[Heer e Bostock](https://hci.stanford.edu/publications/2010/crowd-perception/heer-chi2010.pdf)
investigam estimativas de proporção e a influência do desenho.

**Razão estimada** compara a menor magnitude com a maior: $100v_{menor}/v_{maior}$.
**Erro absoluto** é $|estimativa-valor\ real|$. Se ambos são percentuais, o erro é em pontos
percentuais (p.p.). O **erro logarítmico** usado nesse tipo de experimento pode ser expresso como
$\log_2(|estimativa-valor\ real|+1/8)$: comprime a escala de erros; menor significa melhor.
Neste laboratório usaremos erro absoluto, mais fácil de interpretar.

**Lei de Weber:** em determinadas condições, o menor incremento perceptível $\Delta I$ é
aproximadamente proporcional à intensidade $I$: $\Delta I/I\approx k$. Não é uma regra universal
fora das condições em que foi medida. **Lei de Stevens:** $S=kI^n$ relaciona estímulo físico $I$
à sensação $S$. Expoentes ilustrativos discutidos na aula são aproximadamente 1 para comprimento,
0,7 para área, 0,5–0,6 para volume e 0,33–0,5 para luminosidade. Eles dependem de tarefa e condições;
não são fatores para “corrigir” arbitrariamente um gráfico de dados.

**Raio versus área:** em um círculo, $A=\pi r^2$. Para área proporcional a um valor positivo $v$,
use $r\propto\sqrt v$. Se o raio dobra, a área quadruplica. Esse efeito geométrico é diferente
do viés perceptual descrito por Stevens.

**Bertin:** além das duas dimensões do plano, descreve tamanho, valor/luminosidade, textura, cor,
orientação e forma. Uma variável visual é **seletiva** se ajuda a isolar um grupo; **associativa**
se permite perceber elementos como equivalentes; **ordenada** se comunica sequência; e
**quantitativa** se permite relacionar magnitudes. Mackinlay amplia a discussão de codificações
adequadas a tipos de dados; Munzner a organiza em termos de dados, tarefas, marcas e canais.

**Canais separáveis** podem ser lidos com relativa independência, como posição e cor.
**Canais integrais**, como altura e largura de um retângulo, tendem a se combinar na percepção
de tamanho. **Interferência** ocorre quando um canal dificulta a leitura de outro: cores em pontos
minúsculos podem ser difíceis de distinguir.

**Pizza e rosca:** representam partes de um todo por setores/arcos e áreas. Comparações próximas
podem ser difíceis; barras alinhadas ajudam quando a tarefa é ordenar ou estimar diferenças.
A rosca remove o centro, mas seu desempenho depende da tarefa; não se assume inferioridade
universal. Perspectiva 3D pode deformar áreas aparentes. **Waffle** usa unidades em uma grade;
**barra 100%** divide um comprimento total; **slopegraph** conecta valores entre duas condições.
São alternativas para tarefas específicas, não soluções automáticas.

## Dicionário da aula 03 — Gestalt, hierarquia e cor

**Gestalt** reúne princípios que descrevem como organizamos elementos em conjuntos perceptuais.
Esses agrupamentos podem sugerir relações que não existem nos dados.

| Princípio | Definição | Aplicação e risco |
|---|---|---|
| Proximidade | Elementos próximos parecem pertencer ao mesmo grupo | Aproximar rótulo e série; evitar unir indicadores sem relação |
| Similaridade | Elementos semelhantes parecem relacionados | Manter o significado da mesma cor em todos os painéis |
| Cercamento/região comum | Uma região delimitada sugere um grupo | Destacar um período; evitar caixas em tudo |
| Fechamento | Completamos mentalmente contornos incompletos | Dispensar moldura quando alinhamento já organiza |
| Continuidade | Tendemos a seguir trajetórias suaves | Linhas sugerem continuidade; explicitar lacunas |
| Conexão | Elementos ligados parecem relacionados | Ligar antes/depois do mesmo item; não inventar ordem entre categorias |
| Figura e fundo | Distinguimos objeto principal e contexto | Marcas em destaque, grade discreta |
| Destino comum | Elementos que se movem juntos parecem agrupados | Animação consistente entre estados |

Conexão e região comum podem dominar outros sinais, mas não há um “princípio mais forte” em
todas as situações. **Ausente** significa não observado; **zero** é um valor observado.
Unir pontos separados por ausência pode sugerir uma interpolação. **Interpolar** é estimar
valores entre observações, o que exige regra e identificação explícitas.

**Hierarquia visual** organiza prioridades de leitura por tamanho, peso, cor/contraste e espaço.
**Alinhamento e grade de layout** organizam posições; **espaço em branco** separa grupos e reduz
competição. **Ruído visual** são elementos que dificultam a tarefa sem benefício proporcional.
Os padrões de varredura **Z**, **F** e o **diagrama de Gutenberg** são heurísticas de percurso
do olhar em certos layouts e culturas de leitura, não leis. O **teste dos cinco segundos** pede
a uma pessoa que relate a mensagem após exposição breve: é diagnóstico prático, não certificação.

### Cor: propriedades, espaços e escalas

**Matiz** é a família de cor; **saturação/croma** descreve intensidade cromática, com definições
que variam entre espaços; **luminosidade perceptual** é a aparência de claro/escuro.
**Luminância relativa** é uma grandeza calculada que usaremos no contraste; não é o mesmo
que o parâmetro L do HSL.

**RGB** combina componentes vermelho, verde e azul; **sRGB** especifica um espaço RGB e sua
codificação. **HSL** reorganiza RGB em matiz, saturação e lightness, mas distâncias iguais não
garantem diferenças perceptuais iguais. **CIELAB** aproxima distâncias perceptuais; **LCh/HCL**
usam coordenadas polares de espaços desse tipo. **Oklab/Oklch** são outros modelos para aproximar
uniformidade perceptual. Nenhum dispensa observar o tamanho da marca, o fundo e a tela.

| Escala | Definição | Exemplo e cuidado |
|---|---|---|
| Qualitativa | Cores para categorias sem ordem | Okabe–Ito; manter correspondência categoria/cor |
| Sequencial | Progressão visual de menor a maior | viridis/cividis; apresentar legenda e domínio |
| Divergente | Dois lados de uma referência significativa | Diferença em relação à meta; centro no zero da diferença |
| Cíclica | Extremos se encontram em um ciclo | Horário ou direção; twilight |

**Domínio** é o intervalo de dados mapeado à escala. Em uma divergente, o centro pode ser zero,
uma meta ou outro valor justificável; não precisa ser o meio aritmético dos extremos.
**Arco-íris/jet** pode criar saliências ou faixas perceptuais em um gradiente numérico regular,
pois sua luminosidade não progride uniformemente. **viridis**, magma, inferno e plasma são opções
sequenciais; **cividis** foi projetada considerando deficiências de visão de cores. **ColorBrewer**
organiza paletas por tarefa cartográfica. **Okabe–Ito** é uma paleta qualitativa conhecida por
favorecer distinção de categorias, mas seu nome não garante contraste suficiente contra qualquer fundo.

Cinza com uma cor de acento pode direcionar atenção. Muitas categorias excedem a capacidade de
distinção: cerca de sete cores é uma regra prática, não um limite biológico fixo. Mais cores não
resolvem automaticamente excesso de séries. Ordenação, agrupamento e pequenos painéis podem ajudar.

### Acessibilidade: três perguntas diferentes

**Deficiência de visão de cores (CVD)** altera a discriminação de cores. Protanopia e deuteranopia
envolvem sistemas de cones de comprimentos de onda longos e médios; tritanopia, curtos. As formas
“anomalia” indicam alterações, não ausência completa desse sistema. Acromatopsia envolve restrições
mais amplas à visão de cores. Evitem entender esses nomes como simples filtros que “apagam uma cor”.
As estimativas populacionais variam; a referência de 8% dos homens e 0,5% das mulheres diz respeito
a certas populações de ascendência europeia, não a qualquer turma.

1. **Cor é a única pista?** O critério [WCAG 1.4.1](https://www.w3.org/WAI/WCAG22/Understanding/use-of-color)
   pede outra forma de transmitir a informação: rótulo, forma, padrão ou texto.
2. **Há contraste suficiente?** Para texto normal, [1.4.3](https://www.w3.org/WAI/WCAG22/Understanding/contrast-minimum.html)
   exige ao menos 4,5:1; para texto grande, 3:1. Grande significa ao menos 18 pt regular ou 14 pt em
   negrito (aproximadamente 24 e 18,67 CSS px). Objetos gráficos necessários à compreensão têm
   requisitos de 3:1 contra cores adjacentes em [1.4.11](https://www.w3.org/WAI/WCAG22/Understanding/non-text-contrast),
   consideradas as condições e exceções do critério.
3. **As categorias continuam distinguíveis?** Testem redundância, impressão em cinza e, quando
   possível, simulações de protanopia, deuteranopia e tritanopia. Cinza não simula essas condições;
   uma simulação tampouco substitui usuários reais.

**Contraste:** para cada componente sRGB $c$ normalizado entre 0 e 1, linearize com
$c/12{,}92$ se $c\leq0{,}04045$, ou $((c+0{,}055)/1{,}055)^{2{,}4}$ caso contrário.
Então $L=0{,}2126R+0{,}7152G+0{,}0722B$ e
$CR=(L_{claro}+0{,}05)/(L_{escuro}+0{,}05)$.
Meçam cada elemento em relação ao fundo ou à cor adjacente pertinente; não arredondem para decidir
se o limiar foi atingido. Aprovar esses cálculos não certifica a acessibilidade de todo o notebook.
Em modo escuro, repitam a avaliação com o fundo real.

## Dicionário da aula 04 — design analítico, integridade e ética

**Excelência gráfica**, em Tufte, relaciona conteúdo, clareza e eficiência de comunicação.
**Design analítico** favorece comparações, explicitação de mecanismos quando há evidência,
integração de palavras/números/imagens, documentação das fontes e contexto multivariado.
Desenhar setas ou duas curvas semelhantes não demonstra um mecanismo causal.

**Integridade gráfica** é a correspondência entre representação, dados e afirmação.
**Data-ink ratio** é a razão entre tinta dedicada aos dados e tinta total. É uma heurística:
contar pixels não resolve o que tem função informativa. **Chartjunk** é ornamentação que compete
com a leitura, como perspectiva decorativa, sombras e grades excessivas. Uma hachura que distingue
séries pode ser útil; uma hachura vibratória sem função pode ser ruído. O **“pato”** descreve a
ilustração que passa a dominar a representação.

**Densidade de dados** relaciona quantidade de entradas e área gráfica; aumentar densidade não
garante legibilidade. **Small multiples** são pequenos painéis com desenho repetido e escalas
comparáveis. **Sparklines** são séries compactas junto ao texto; precisam de contexto suficiente
para não esconder magnitude, intervalo ou referência.

**O contraditório:** a discussão de Bateman et al., *Useful Junk?* (2010), presente na aula,
mostra por que não se deve supor que retirar toda ilustração melhora qualquer objetivo.
Compreensão, preferência e memória são resultados distintos. Estudos de preferência como o de
Inbar et al. também não tornam decoração sempre desejável. A tarefa e a evidência de uso devem
orientar a edição: remover unidade, fonte ou referência pode piorar o gráfico mesmo com menos tinta.

### Lie factor: efeito relativo no desenho dividido pelo efeito relativo nos dados

Para dois valores positivos distintos $d_1,d_2$ e suas medidas visuais correspondentes $g_1,g_2$:

$$E_d=\frac{d_2-d_1}{d_1},\quad E_g=\frac{g_2-g_1}{g_1},\quad LF=\frac{E_g}{E_d}.$$

Se medimos barras, $g$ é comprimento visível; em círculos que codificam área, $g$ é a área,
não o raio. **LF próximo de 1** indica proporcionalidade para aquele efeito; acima de 1, amplificação;
entre 0 e 1, atenuação. A faixa didática 0,95–1,05 não é certificado de honestidade: um gráfico
proporcional ainda pode omitir grupos, fontes ou contexto. Não ordenem valores e medidas de modo
independente; preservem a correspondência. Inversões de sinal precisam de diagnóstico próprio.
Com base zero ou efeito real zero, a razão pode ser indefinida; usem diferença absoluta e descrevam
o problema. Não inventem um LF para todo tipo de defeito.

**Exemplo externo ao exercício:** dados 50→60 têm efeito de 20%; comprimentos 10→20 têm efeito
de 100%; LF=5. A segunda barra tem o dobro do comprimento, mas o valor não dobrou.

| Distorção discutida na aula | Definição | Como investigar/corrigir |
|---|---|---|
| Eixo truncado em barras | Cortar a base altera comprimentos relativos | Mostrar zero em barras de magnitude; para detalhes, considerar pontos/linhas e explicitar limites |
| Área/volume desproporcional | Dimensão linear cresce como o dado, inflando área/volume | Área proporcional: raio ∝ raiz quadrada; volume proporcional: dimensão ∝ raiz cúbica |
| Duplo eixo Y | Duas escalas independentes permitem alinhamentos visuais arbitrários | Separar painéis ou usar índice base 100 quando adequado; manter unidades explícitas |
| Eixo invertido | Valores crescem na direção oposta à expectativa | Declarar direção e justificar convenção, como profundidade |
| Recorte conveniente | Selecionar janela que favorece a narrativa | Exibir período completo disponível e justificar a janela |
| Agregação enganosa | Total/média esconde composição ou subgrupos | Mostrar denominadores, ponderar e desagregar conforme a pergunta |
| Escala log não comunicada | Razões iguais ocupam distâncias iguais sem explicar isso | Rotular e explicar; log usual não admite zero ou negativos |

**Escala linear** dá distâncias iguais a diferenças iguais. **Escala logarítmica** dá distâncias
iguais a razões iguais: 10→100 e 100→1000 ocupam o mesmo intervalo. Ela pode ser apropriada;
o problema é deixar o leitor interpretar sua geometria como linear. **Base 100** transforma
$x_t$ em $100x_t/x_0$ quando $x_0\ne0$: mostra variação relativa, perdendo a unidade original.
Modificar limites de eixos altera a aparência, mas **não altera a correlação calculada dos mesmos
pares de valores**. Uma razão de inclinações de janelas diferentes não é, por si só, o LF de Tufte.

Outros cuidados: datas irregularmente espaçadas precisam de distâncias temporais coerentes;
percentuais exigem denominador; dinheiro nominal deve ser distinguido de dinheiro corrigido
por inflação; ordenação alfabética pode esconder padrões; 3D pode introduzir oclusão/perspectiva.
Uma marca bidimensional não é proibida para um valor: o problema é escalar duas dimensões
linearmente pelo mesmo valor e depois deixar a área comunicar uma magnitude inflada.

### Resumos, agregação e limites da conclusão

**Observação** é uma unidade registrada. **População** é o conjunto sobre o qual se deseja concluir;
**amostra** é um subconjunto observado. Dados sintéticos não são amostra de uma população real.
**Média** é soma dividida pela quantidade; **variância amostral** é a soma dos desvios quadráticos
em relação à média dividida por $n-1$; **desvio-padrão** é a raiz da variância.
**Distribuição** descreve como os valores se espalham, incluindo concentração, dispersão e formato.
**Outlier** é uma observação que se destaca do padrão; precisa ser investigada, não apagada
automaticamente. Um ponto **influente** pode alterar fortemente um resumo ou uma reta ajustada.

**Correlação de Pearson ($r$)** resume associação linear entre duas variáveis, de −1 a +1;
não captura toda relação não linear e pode ser sensível a pontos influentes. Uma **reta ajustada**
resume uma tendência linear; o código a calcula apenas para comparação visual, sem exigir estudo
de regressão. **Anscombe** construiu quatro conjuntos com resumos aproximadamente iguais e formas
muito diferentes. Pequenas diferenças numéricas decorrem da precisão publicada dos valores.
[Datasaurus Dozen](https://www.research.autodesk.com/publications/same-stats-different-graphs/)
estende essa ideia com treze desenhos e resumos iguais até duas casas decimais. O exercício usa
Anscombe; não precisamos baixar o Datasaurus para compreender o argumento.

**Contagem** é número de ocorrências. **Taxa/proporção** é ocorrências divididas pelo total elegível.
Para grupos mutuamente exclusivos, a taxa agregada é
$\sum sucessos/\sum total$ — uma média das taxas ponderada pelos tamanhos dos grupos.
A média simples das taxas responde à atribuição de pesos iguais, que é outra pergunta.
**Pontos percentuais** medem subtração de percentuais: 40%→50% são +10 p.p.
**Variação relativa** divide a diferença pelo valor inicial: o mesmo caso é +25%.

**Paradoxo de Simpson** é a inversão de uma associação ao agregar ou separar grupos, devido à
composição e aos pesos. Em situações mais gerais, agregar também pode esconder ou enfraquecer
associações sem inverter todos os subgrupos. O caso Berkeley é a referência histórica da aula;
aqui usamos dados fictícios mais simples para tornar os pesos verificáveis.

**Associação** é um padrão conjunto; **causalidade** significa que intervir em algo modifica o
resultado, sob condições definidas. **Confundimento** ocorre quando outra variável afeta a
interpretação da relação investigada. Associação pode envolver causa, causalidade reversa,
fatores comuns ou coincidência. **Correlação espúria** pode aparecer sem o mecanismo causal
alegado, inclusive por tendências temporais comuns.

**Desagregar não prova causalidade nem ausência de discriminação.** Escolher variáveis de controle
exige conhecimento do processo: nem toda variável deve ser controlada. Um desenho observacional
limitado não permite afirmar o que aconteceria com o mesmo caso sob outra equipe.

**Ética na visualização** inclui declarar fonte, recorte, unidade, denominador, transformações,
incertezas e limitações. **Incerteza** é o que não sabemos sobre medida, interpretação ou
generalização; não precisa ser convertida em intervalo estatístico sem dados e método adequados.
**Reprodutibilidade** significa conseguir repetir a análise a partir dos mesmos dados e instruções.
**Auditoria de IA** confronta cada afirmação gerada com cálculos e fontes; fluência não comprova
correção. Pode-se auditar sem contratar ou acessar um modelo, usando um texto didático fornecido.

```python
import sys
import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
from matplotlib.patches import Circle
from matplotlib.colors import to_rgb, Normalize, TwoSlopeNorm
from IPython.display import display

plt.rcParams.update({
    "figure.dpi": 110, "font.size": 11, "axes.titlesize": 12,
    "axes.labelsize": 11, "axes.spines.top": False,
    "axes.spines.right": False, "figure.facecolor": "white",
    "axes.facecolor": "white", "text.color": "#222222",
    "axes.labelcolor": "#222222",
})
AZUL, LARANJA, CINZA = "#0072B2", "#A64700", "#666666"

def finalizar(fig, descricao):
    '''Mostra figura e descrição textual; os dados ficam disponíveis em tabelas.'''
    fig.tight_layout()
    plt.show()
    print("Descrição da figura:", descricao)

def luminancia(cor):
    rgb = np.array(to_rgb(cor))
    linear = np.where(rgb <= 0.04045, rgb / 12.92, ((rgb + 0.055) / 1.055) ** 2.4)
    return float(linear @ np.array([0.2126, 0.7152, 0.0722]))

def contraste(cor, fundo="#FFFFFF"):
    a, b = sorted([luminancia(cor), luminancia(fundo)])
    return (b + 0.05) / (a + 0.05)

def cinza_luminancia(cor):
    l = luminancia(cor)
    c = 12.92*l if l <= 0.0031308 else 1.055*l**(1/2.4)-0.055
    return (c, c, c)

print("Ambiente:", sys.version.split()[0], "| NumPy", np.__version__,
      "| pandas", pd.__version__, "| Matplotlib", matplotlib.__version__)
print("Preparação concluída. Próximo passo: leiam a hipótese da investigação 1.")
```

### Conferência numérica — não substitui a justificativa

Comparem com as contas anteriores. A função abaixo só trata bases positivas e efeitos diferentes de zero; mantém a correspondência entre o primeiro dado e a primeira medida visual.

```python
def fator_distorcao(dados, medidas):
    d1, d2 = dados
    g1, g2 = medidas
    if min(d1, d2, g1, g2) <= 0 or d1 == d2:
        raise ValueError("Este cálculo exige bases/medidas positivas e efeito real diferente de zero.")
    ed, eg = (d2-d1)/d1, (g2-g1)/g1
    return {"efeito nos dados (%)": 100*ed, "efeito visual (%)": 100*eg, "LF": eg/ed}

lf_barras = fator_distorcao([94, 98], [94-90, 98-90])
lf_zero = fator_distorcao([94, 98], [94, 98])
display(pd.DataFrame([lf_barras, lf_zero], index=["Base 90", "Base zero"]).round(4))

fig, axes = plt.subplots(1, 2, figsize=(11, 3.5))
for ax, raios, titulo in [(axes[0], [.5, 1], "Raio proporcional ao valor"),
                          (axes[1], [.5, .5*np.sqrt(2)], "Área proporcional ao valor")]:
    for x, raio, valor in zip([1.2, 3.6], raios, [10, 20]):
        ax.add_patch(Circle((x, 1.2), raio, color=AZUL))
        ax.text(x, -.05, f"{valor} mil", ha="center")
    ax.set(xlim=(0, 5), ylim=(-.3, 2.5), aspect="equal", title=titulo)
    ax.axis("off")
finalizar(fig, "Em ambos os painéis, círculos para 10 e 20 mil. À esquerda o raio dobra; à direita o raio é multiplicado pela raiz quadrada de dois.")
lf_area_errada = fator_distorcao([10, 20], [np.pi*.5**2, np.pi*1**2])
lf_area_correta = fator_distorcao([10, 20], [np.pi*.5**2, np.pi*(.5*np.sqrt(2))**2])
display(pd.DataFrame([lf_area_errada, lf_area_correta],
                     index=["Raio proporcional", "Área proporcional"]).round(4))
```
